# importing libraries

```python
import pandas as pd
import numpy as np
```

# loading the dataset

```python
data = pd.read_csv("KNN for missing year.csv")
data.head()
```

```python
data.describe()
```

```python
data['Project Completion Date']
```

# Converting the project completion datatype

```python
data['Project Completion Date'] = pd.to_datetime(data['Project Completion Date'], format = "%Y-%m-%d")
```

```python
import datetime as dt
data['Project Completion Year'] = data['Project Completion Date'].dt.year
data['Project Completion month'] = data['Project Completion Date'].dt.month
data['Project Completion Day'] = data['Project Completion Date'].dt.day
```

```python
data.drop(columns="Project Completion Date" , inplace= True)
```

```python
data.head()
```

```python
data.dtypes
```

# Target variable

```python
data['Year Home Built'] = data['Year Home Built'].replace("<1800", 1790)
data['Year Home Built'] = pd.to_numeric(data['Year Home Built'])
```

# Filtering categorical columns 

```python
cat_col = []
num_col =[]
for i in data.columns:
    if data[i].dtypes == "object":
        cat_col.append(i)
    else:
        num_col.append(i)
```

```python
cat_col_df = data[cat_col]
cat_col_df
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
for col in cat_col_df.columns:
    # fit_transform handles one column at a time
    cat_col_df[col] = le.fit_transform(cat_col_df[col])
```

```python
cat_col_df
```

```python
data['Job Type'] = cat_col_df['Job Type']
data['Type Of Dwelling'] = cat_col_df['Type Of Dwelling']
data['Pre-Retrofit Home Heating Fuel Type'] = cat_col_df['Pre-Retrofit Home Heating Fuel Type']
```

```python
data.head()
```

```python
data.dtypes
```

# Standard Scaling

```python
from sklearn.preprocessing import StandardScaler

StandardScaler = StandardScaler()
data_scaled = StandardScaler.fit_transform(data)
```

```python
data_scaled
```

```python
num_col_df_scaled = pd.DataFrame(data_scaled, columns=data.columns)

num_col_df_scaled.head()
```

# Imputing Year

```python
from sklearn.impute import KNNImputer

year_imputer = KNNImputer(n_neighbors= 5, weights="distance")
knn_year_imputed = year_imputer.fit_transform(data_scaled)
```

```python
knn_year_imputed = StandardScaler.inverse_transform(knn_year_imputed)
knn_year_imputed_df = pd.DataFrame(knn_year_imputed, columns=data.columns, index=data.index)
```

```python
knn_year_imputed_df
```

```python
knn_year_imputed_df["Year Home Built"] = knn_year_imputed_df["Year Home Built"].round()
```

```python
knn_year_imputed_df.isnull().sum()
```

```python
knn_year_imputed_df["Year Home Built"] = knn_year_imputed_df['Year Home Built'].astype(int)
```

```python
data = knn_year_imputed_df
data.head()
```
